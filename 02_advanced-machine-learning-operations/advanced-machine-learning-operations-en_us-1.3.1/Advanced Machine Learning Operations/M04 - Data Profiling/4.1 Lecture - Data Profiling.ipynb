# Lecture - Data Profiling

Once a model is in production, monitoring keeps it trustworthy. This lecture distinguishes the types of monitoring and drift — data, concept, model-quality, and bias — and contrasts siloed classical monitoring with a unified modern approach. You will see what Databricks Data Profiling measures across snapshot, time-series, and inference tables, how to add custom metrics and SQL alerts, and how to visualize results in AI/BI dashboards and orchestrate them with Lakeflow Jobs.

            <ul style="padding-left: 30px;"><li>Ensuring the ML solution <b>aligns with and fulfills specific business objectives</b>.</li></ul><br>
        <li><b>Resource Utilization:</b></li>
            <ul style="padding-left: 30px;"><li>Ensuring <b>efficient resource utilization</b> within the ML infrastructure.</li></ul><br>
        <li><b>Traceability:</b></li>
            <ul style="padding-left: 30px;"><li><b>Tracking data lineage</b> to understand the origin, movement, and transformation of data.</li></ul><br>
        <li><b>Model Performance:</b></li>
            <ul style="padding-left: 30px;"><li>Detecting and addressing types of <b>drift or degradation</b>.</li></ul>
        </li>
        </ul>
    </td>
    <td style="padding-top: 10px; font-size: 1em;">
        <ul>
        <li><b>Data Drift:</b></li>
            <ul style="padding-left: 30px;"><li>Occurs when the <b>statistical properties</b> of the input data <b>change over time</b>.</li></ul><br>
        <li><b>Concept Drift:</b></li>
            <ul style="padding-left: 30px;"><li>Happens when the <b>relationship</b> between input features and the <b>target variable</b> changes.</li></ul><br>
        <li><b>Model Quality Drift:</b></li>
            <ul style="padding-left: 30px;"><li>Reflects a <b>decrease in the model's predictive performance</b> over time.</li></ul><br>
        <li><b>Bias Drift:</b></li>
            <ul style="padding-left: 30px;"><li>Involves shifts in model outcomes that could lead to <b>unfair treatment</b> of certain groups.</li></ul>
        </li>
        </ul>
    </td>
  </tr>
</table>

            '</div>';

        var codeEl = document.getElementById(id);
        codeEl.textContent = code;
        Prism.highlightElement(codeEl);

        block.querySelector('.copy-btn').onclick = function() {
            var t = document.createElement('textarea');
            t.value = code;
            document.body.appendChild(t);
            t.select();
            document.execCommand('copy');
            document.body.removeChild(t);
            this.textContent = '✓ Copied!';
            setTimeout(() => this.textContent = 'Copy', 2000);
        };
    });
})();
