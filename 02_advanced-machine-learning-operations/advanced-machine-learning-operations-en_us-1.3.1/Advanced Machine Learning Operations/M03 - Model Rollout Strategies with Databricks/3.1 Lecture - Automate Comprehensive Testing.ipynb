## F. Testing notebooks Using Testing Frameworks

- [<span style="color:#e5302a;">**Pytest**</span>](https://docs.pytest.org/en/stable/)
    - Popular testing framework for Python.
    - Supports fixtures, parametrized tests, and built-in assertions.
    - Works well with Databricks Python notebooks by importing functions for isolated tests.

- [<span style="color:#e5302a;">**Testthat**</span>](https://testthat.r-lib.org/)
    - Primarily used for R notebooks.
    - Allows robust unit testing with easy-to-write test cases.
    - Supports reproducibility and detailed error messages for R code.

- [<span style="color:#e5302a;">**ScalaTest**</span>](https://www.scalatest.org/)
    - Widely used for testing Scala code in Databricks notebooks.
    - Supports behavior-driven and test-driven development.
    - Provides support for both unit testing and integration testing of Spark jobs.

- [<span style="color:#e5302a;">**SQL**</span>](https://docs.databricks.com/aws/en/notebooks/testing#sql)
    - SQL queries can be validated using assertions in notebooks.
    - Useful for testing data transformations, aggregations, and schema validation.
    - Ensures query outputs match expected results in unit tests.


## G. Optimize Tests in the CI/CD Process

### Balancing Speed, Cost, and Comprehensiveness

Running every test on every change at every stage of the pipeline is inefficient. Aiming for "zero defects in production" leads to diminishing returns and poor ROI. Instead, be **selective**: prioritize high-value tests, catch defects as early as possible, and keep the ability to detect and roll back issues that reach production.

- **"Zero-defect in production" is unrealistic.** Prioritize catching critical defects while accepting that minor, low-impact issues may occur.
- **Use a defect funnel — catch defects early.** The later a defect is detected, the more expensive it is to handle. Rely on unit and integration tests to address issues before they escalate.
- **Cover what earlier stages miss.** Add end-to-end and scenario-based tests to capture defects that cannot be caught in earlier stages.
- **Detect and roll back in production.** Ensure workflows can detect and mitigate issues quickly using approaches such as **canary deployments**, **probing and monitoring**, and **controlled experiments**.
- **Be selective and data-driven.** Automate the majority of tests, reserve manual testing for critical workflows, and use analytics to remove redundant, low-value tests.
