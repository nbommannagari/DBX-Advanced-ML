# Lecture - Streamlining Development to Deployment

Great MLOps starts with sound architecture. This lecture introduces the guiding principles for organizing ML environments and moving work from development to production. You will compare direct and indirect environment separation, understand why isolated development, staging, and production workspaces are recommended, and explore why teams promote code rather than models between environments — along with the deploy-model, deploy-code, and hybrid patterns and their trade-offs.

        <div style="height:8px;"></div>
        <li><b>Simpler environments</b></li>
            <ul><li>Easier to manage <b>security</b> and <b>compliance</b></li>
            <li>More straightforward to <b>track</b> changes and issues</li></ul>
        <div style="height:8px;"></div>
        <li><b>Reduces the risk</b> of cross-environment interactions</li>
        <div style="height:8px;"></div>
        <li><b>Scales</b> well to multiple projects</li>
            <ul><li>Might require more <b>administrative effort</b> to manage multiple workspaces</li>
            <li>Might lead to <b>higher costs</b> due to the need for separate resources for each workspace</li></ul>
        </ul>
    </td>
    <td style="font-size: 14px;">
        <ul>
        <li><b>One Databricks workspace</b> with enforced separation using <b>naming conventions</b> and <b>access control</b></li>
        <div style="height:8px;"></div>
        <li><b>Simpler</b> overall infrastructure</li>
            <ul><li><b>Less permissions</b> required</li>
            <li>Can be more <b>cost-effective</b> needing fewer resources</li></ul>
            <div style="height:8px;"></div>
        <li>Easier cross-environment <b>collaboration</b> and sharing</li>
        <div style="height:8px;"></div>
        <li><b>Doesn’t scale well</b> to multiple projects</li>
            <ul><li><b>Complex</b> individual environment</li>
            <li>Requires management to <b>maintain</b> the separation and prevent accidental cross-environment interactions</li>
            <li>Can be more challenging to manage <b>security</b> and <b>compliance</b> requirements</li></ul>

## G. ML Operations Guiding Principles

### Optimizing Development and Deployment

<br>

- Automated and Orchestrated **Continuous Integration and Continuous Deployment** 
    - Set up CI/CD pipelines to automate the machine learning lifecycle.
    - Reducing manual steps and ensuring consistency.
    - Break down ML workflows into manageable modules.

- Automate Comprehensive **Testing**
    - Implement a testing strategy including: unit, integration, and end-to-end tests.
    - Incorporate canary tests and blue/green deployment with CI/CD Pipelines.

- **Monitoring** and **Alerts**
    - Monitor model performance and operations with visuals, custom metrics, and insights.
    - Set up automated alerts to detect issues early.

- Build ML **Assets as Code**
    - Standardize and automate ML asset creation using Infrastructure as Code and predefined templates.
