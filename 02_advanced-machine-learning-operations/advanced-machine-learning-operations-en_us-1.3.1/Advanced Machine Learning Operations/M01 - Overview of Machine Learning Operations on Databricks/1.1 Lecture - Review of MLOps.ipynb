# Lecture - Review of MLOps

MLOps brings the discipline of software operations to machine learning. This opening lecture reviews how Databricks supports the entire ML lifecycle — from data and training to serving and monitoring — and shows how the three operational pillars of DataOps, DevOps, and ModelOps combine to manage data, code, and models reliably at scale. It frames the development and production workflows that the rest of the course builds on.

## E. DataOps, DevOps, and ModelOps are…

### A set of practices, processes, and technologies to: 

<div style="display:flex;gap:0;margin:24px 0;align-items:stretch;">

  <div style="flex:1;background:#FFF3E0;border-top:4px solid #F57C00;padding:24px;">
    <div style="font-weight:bold;color:#E65100;font-size:18px;margin-bottom:14px;">DataOps</div>
    <u>Ensure data <b>quality</b></u>
    <div style="height:8px;"></div>
    <ul style="font-size:16px;color:#555;line-height:2;margin:0;padding-left:20px;">
        <li>Optimize data <b>processing</b></li>
        <li>Centralize data <b>discovery</b>, <b>management</b>, and <b>governance</b></li>
        <li>Establish traceable data <b>lineage</b> and <b>monitoring</b></li>
        <li><b>Enhance collaboration</b> across teams</li>
        <li><b>Monitor</b> data</li>
    </ul>
  </div>

  <div style="flex:1;background:#E3F2FD;border-top:4px solid #1976D2;padding:24px;">
    <div style="font-weight:bold;color:#0D47A1;font-size:18px;margin-bottom:14px;">DevOps</div>
    <u>Treat machine learning as <b>code</b></u>
    <div style="height:8px;"></div>
    <ul style="font-size:16px;color:#555;line-height:2;margin:0;padding-left:20px;">
        <li>Automate <b>CI/CD</b></li>
        <li>Enable continuous <b>code testing</b></li>
        <li><b>Version control</b></li>
        <li>Establish production-grade <b>workflows</b></li>
            <ul style="font-size:14px;"><li>Orchestration & automation</li></ul>
        <li><b>Monitor</b> system performance</li>
    </ul>
  </div>

  <div style="flex:1;background:#E8F5E9;border-top:4px solid #388E3C;border-radius:0px 0 0 0px;padding:24px;">
    <div style="font-weight:bold;color:#1B5E20;font-size:18px;margin-bottom:14px;">ModelOps</div>
    <u>Move beyond models as <b>objects</b></u>
    <div style="height:8px;"></div>
    <ul style="font-size:16px;color:#555;line-height:2;margin:0;padding-left:20px;">
        <li>Treating <b>model code</b> as <b>software</b></li>
        <li>Treating models as <b>data</b></li>
        <li>Manage the <b>model lifecycle</b></li>
        <li><b>Monitor</b> model performance</li>
