# Lecture - Build ML assets as Code

Treating infrastructure as code makes ML projects reproducible, versioned, and easy to deploy everywhere. This lecture breaks down the anatomy of an ML project — its code, resources, and execution environment — and shows how Declarative Automation Bundles (DABs) describe all of it in YAML. It then contrasts traditional MLOps stacks with Databricks MLOps Stacks, a customizable framework that scaffolds CI/CD, infrastructure-as-code, and project structure for productionizing ML.

## A. Infrastructure as Code

<br>

- The way to describe infrastructure objects using general purpose or specialized language

- Pros:
    - **Easy to create new** environments & maintain them
    - **Versioned**, easy to rollback in case of problems
    - **Changes** are made via pull requests, that are **tested** by build pipelines, **published** by release pipelines to multiple environments (staging/production)
    - Could help to build **completely automated releases**, without “manual” access to production environments
    - Usually it’s easy to **integrate with CI/CD** pipelines

- Popular implementations: **Declarative Automation Bundles**, HashiCorp Terraform, Ansible, AWS CloudFormation, Azure Resource Management templates, ...
