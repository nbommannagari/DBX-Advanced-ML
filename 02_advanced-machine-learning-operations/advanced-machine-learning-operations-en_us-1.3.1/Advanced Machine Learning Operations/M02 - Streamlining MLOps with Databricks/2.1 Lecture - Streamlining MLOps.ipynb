# Lecture - Streamlining MLOps

Modern MLOps is built on four continuous practices. This lecture defines Continuous Integration, Continuous Training, Continuous Deployment, and Continuous Monitoring, and maps each to the concrete pipelines that implement it. You will see how these practices fit together into a recommended end-to-end architecture and which core Databricks tools — Lakeflow Jobs, Managed MLflow, and Git folders — make them work.

## A. Continuous Terminology

### <span style="color:#4d7dad;"><b>DevOps</b></span>, <span style="color:#ed5e47;"><b>Data Ops</b></span>, <span style="color:#4ba973;"><b>Model Ops</b></span>

<br>

- **Continuous Integration (CI)** extends the testing and validating code by adding testing and validating of <span style="color:#4d7dad;"><b>code</b></span>, <span style="color:#ed5e47;"><b>data</b></span> and <span style="color:#4ba973;"><b>models</b></span>.

- **Continuous Training (CT)** <span style="color:#4d7dad;"><b>automatically</b></span> (re)trains ML <span style="color:#4ba973;"><b>models</b></span> on <span style="color:#ed5e47;"><b>data</b></span> for (re)deployment. 

- **Continuous Deployment (CD)** concerned with <span style="color:#4d7dad;"><b>delivery</b></span> of a ML training pipeline that <span style="color:#4d7dad;"><b>automatically</b></span> deploys another ML <span style="color:#4ba973;"><b>model</b></span> and needed <span style="color:#ed5e47;"><b>stores</b></span>.

- **Continuous Monitoring (CM)** concerned with <span style="color:#4d7dad;"><b>continuous monitoring</b></span> of <span style="color:#ed5e47;"><b>production data</b></span> and <span style="color:#4ba973;"><b>model performance metrics</b></span>.

<hr>

<div style="background-color: #1b5162; color: #fff; padding: 12px 16px; border-radius: 4px; margin: 8px 0;">
  <b>Goal</b>: Increase developer productivity, improve model quality, code quality, and accelerate trained model delivery.
</div>

<br>

   <img src="https://files.training.databricks.com/binder/prod_main/advanced-machine-learning-operations-en_us-1.3.1/images/20261023T141558Z/Advanced Machine Learning Operations/Includes/images/M2L1_28.png" alt="Image description" style="max-width: 100%; height: auto;">

  <div style="flex:1;background:#1b5162;border-top:4px solid #000;border-radius:12px 0 0 12px;padding:24px;">
    <div style="font-weight:bold;color:#fff;font-size:16px;margin-bottom:14px;">Continuous Integration (CI)</div>
    <ul style="font-size:16px;color:#fff;line-height:2;margin:0;padding-left:20px;">
        <li>Automatically trigger pipelines after code commits.</li>
        <li>Run tests and validation on code, data, and models.</li>
        <li>Track all changes via version control.</li>
    </ul>
  </div>

  <div style="flex:1;background:#757575;border-top:4px solid #000;border-radius:0 12px 12px 0;padding:24px;">
    <div style="font-weight:bold;color:#fff;font-size:16px;margin-bottom:14px;">Continuous Training (CT)</div>
    <ul style="font-size:16px;color:#fff;line-height:2;margin:0;padding-left:20px;">
        <li>Retrain models automatically when new code or data is available or performance indicates.</li>
        <li>Log all experiments and model metrics for reproducibility.</li>
        <li>Automate hyperparameter tuning.</li>
        <li>Evaluate and validate model.</li>
    </ul>
  </div>

</div>

    <div style="font-weight:bold;color:#fff;font-size:16px;margin-bottom:14px;">Continuous Deployment (CD)</div>
    <ul style="font-size:16px;color:#fff;line-height:2;margin:0;padding-left:20px;">
        <li>Validate models in staging before production.</li>

    </ul>
  </div>

  <div style="flex:1;background:#9e9e9e;border-top:4px solid #000;border-radius:0 12px 12px 0;padding:24px;">

    <ul style="font-size:16px;color:#fff;line-height:2;margin:0;padding-left:20px;">
        <li>Track model performance and drift in production.</li>
        <li>Automate retraining or notifications based on performance metrics.</li>
    </ul>

## I. Core Databricks ML CI/CT/CD/CM Tools

### Automating and Streamlining Machine Learning Pipelines


- **Lakeflow Jobs**
    - Jobs
        - Automates the scheduling and orchestration of ML tasks across different stages.
    - Lakeflow Spark Declarative Pipelines
        - Manages ETL pipelines for real-time data, ensuring fresh data for model training.

- **Managed MLflow**
    - Tracks experiments, manages models, and automates deployment across environments.

- **Databricks Git Folder**
    - Enables version control of code and notebooks for collaborative development.
